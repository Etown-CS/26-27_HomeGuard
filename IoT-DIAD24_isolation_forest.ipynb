```python
# For ===== Recon (442,158 flows) =====
# Scanner: one or two devices with a large extra_ports() (hundreds or thousands), often with few dst_ips, 
#          because a vulnerability scan hits many ports on a few targets.
# No scanner: every device’s numbers are close to its benign numbers. Then the scan isn’t visible at the flow level, 
#             and Recon should go in DROP like MITM.
    
def port_spread(df):
    # skips the gateway .1
    local = df[df["Src IP"].str.startswith("192.168.137.") & (df["Src IP"] != "192.168.137.1")]
    return local.groupby("Src IP").agg(flows=("Dst Port", "size"),
                                       ports=("Dst Port", "nunique"),
                                       dst_ips=("Dst IP", "nunique"))

r = port_spread(recon)
b = port_spread(benign24)
cmp = r.join(b, rsuffix="_benign").fillna(0)

# extra_ports puts at the top the devices that suddenly reach many new ports.
cmp["extra_ports"] = cmp["ports"] - cmp["ports_benign"]
cmp.sort_values("extra_ports", ascending=False).head(10)

# .178 is also the attacker in the SqlInjection file. That fits a lab setup where one attacker machine runs several attacks, 
# which makes it more believable as the scanner here.
# .212 reached about 1,000 ports spread across 204 hosts, so it’s sweeping the network rather than probing a few targets in depth.
```

```python
# Web-Based attack

# Web attacks don’t spread across many ports like a scan does. 
# The attacker sends many HTTP requests to a few web servers on port 80,8080, etc.

# The web-attack signature is a local device sending HTTP flows to other local devices (local web servers), 
# which doesn’t happen much in benign traffic. 
# In the profiles, the following fit that pattern:
# .178 --> .217 / .4 / .86 : 80 
# .147 --> .11 / .13 / .236 : 80 

WEB_PORTS = [80, 443, 8080]

def local_web_clients(df):
    # local device -> local web server (not the gateway); background web traffic goes to external servers
    is_local_src = df["Src IP"].str.startswith("192.168.137.") & (df["Src IP"] != "192.168.137.1")
    is_local_dst = df["Dst IP"].str.startswith("192.168.137.") & (df["Dst IP"] != "192.168.137.1")
    web = df[is_local_src & is_local_dst & df["Dst Port"].isin(WEB_PORTS)]
    return web.groupby("Src IP").agg(flows=("Dst IP", "size"),
                                     dst_ips=("Dst IP", "nunique"))

b_web = local_web_clients(benign24)

for name, d in [("SqlInjection", web_sql_inj),
                ("Uploading_Attack", web_upload_attack),
                ("XSS", web_xss)]:
    cmp = local_web_clients(d).join(b_web, rsuffix="_benign").fillna(0)
    cmp["extra_flows"] = cmp["flows"] - cmp["flows_benign"]
    print(f"=== {name} ===")
    print(cmp.sort_values("extra_flows", ascending=False).head(5), "\n")
```

### Cleaning background traffic
Each attack CSV file has a mix of the real attack and normal background traffic from the same network.
Cleaning keeps only the attack flows.

#### Step 1: Learn what normal traffic looks like

```python
# Destinations that normal devices contact regularly
bg_counts = benign24[["Dst IP", "Dst Port"]].value_counts()
print(bg_counts.head())

# Builds a list of every (Dst IP, Dst Port) pair that benign devices contact at least 20 times
# Any flow in an attack file going to one of these is treated as background and removed.
bg_keys = bg_counts[bg_counts >= 20].index      
print(len(bg_keys), "background destinations")

EXEMPT = {("Spoofing", "DNS_Spoofing")}         # DNS spoofing targets port 53 itself

def drop_background(d, key):
    """Remove flows going to destinations that benign devices use all the time."""
    if key in EXEMPT:
        return d
    is_bg = pd.MultiIndex.from_frame(d[["Dst IP", "Dst Port"]]).isin(bg_keys)
    return d[~is_bg]
```

#### Step 2: Find the huge non-TCP/UDP flows

```python
# ICMP and GRE floods are recorded as Protocol 0 and appear as a few very large flows. 
base0 = benign24[benign24["Protocol"] == 0]

# The threshold is the size that only 1% of benign Protocol-0 flows exceed, and anything larger is kept as attack traffic.
p0_thresh = base0["Total Fwd Packet"].quantile(0.99)
print("proto-0 packet threshold (benign p99):", p0_thresh)

big_p0 = lambda d: (d["Protocol"] == 0) & (d["Total Fwd Packet"] > p0_thresh)
```

### Step 3: A rule for each attack file

each file gets the most precise rule available.

- **Victim rules**: for floods where profile() showed one target getting hammered. So for flood attacks, we kept flows directed at the victim hosts that received most of the traffic in each capture.
- **big_p0**: for the ICMP and Mirai floods.
- **Attacker rules**: for scans and web attacks, where one source did the attacking.
- **drop_background**: the fallback for BruteForce and DNS_Spoofing, which have no rule of their own.
- **DROP**: MITM_ArpSpoofing is skipped entirely.

```python
IP = lambda *xs: [f"192.168.137.{x}" for x in xs]

RULES = {   
    # Group A: clear victims (from profile() output)
    ("DoS", "SYN_Flood"):          lambda d: d["Dst IP"].isin(IP(187, 63)),
    ("DoS", "UDP_Flood"):          lambda d: d["Dst IP"].isin(IP(160, 235, 218)),
    ("DoS", "HTTP_Flood"):         lambda d: d["Dst IP"].isin(IP(211, 131, 235, 71, 64)),
    ("DDoS", "HTTP_Flood"):        lambda d: d["Dst IP"].isin(IP(132, 139)),
    ("DDoS", "ACK_Fragmentation"): lambda d: d["Dst IP"].isin(IP(132, 120)) | (d["Src IP"] == IP(123)[0]),
    # Group B: floods hidden in Protocol 0
    ("DDoS", "ICMP_Flood"):         big_p0,
    ("DDoS", "ICMP_Fragmentation"): big_p0,
    ("Mirai", "Greeth_Flood"):      big_p0,
    # Group C: attackers visible in port_spread() output
    ("Recon", "VulnerabilityScan"):    lambda d: d["Src IP"].isin(IP(178, 212)),
    # Group D: attackers visible in local_web_clients() output
    ("Web-Based", "SqlInjection"):     lambda d: d["Src IP"].isin(IP(178)),
    ("Web-Based", "Uploading_Attack"): lambda d: d["Src IP"].isin(IP(147)),
    ("Web-Based", "XSS"):              lambda d: d["Src IP"].isin(IP(147)),
}

DROP = {("Spoofing", "MITM_ArpSpoofing")}   # ARP isn't IP, so no usable flows

def clean_attack(d, key):
    if key in RULES:
        return d[RULES[key](d)]
    return drop_background(d, key)          # BruteForce, DNS_Spoofing
```

```python
# set the max number of rows for each attack file
CAP = 10000
META_KEEP = ["Src IP", "Dst IP", "Timestamp"]   # metadata for analysis, not model input
MIN_FLOWS = 20                                  # skip files with too few flows to split
val_parts, test_parts = [], []

for (cat, sub), d in attacks.items():
    if (cat, sub) in DROP:
        continue
        
    # clean_attack() runs before selecting columns, because 
    # the rules need Protocol and Dst Port, and those are no longer in feature_cols.
    d = clean_attack(d, (cat, sub))             # keep only attack traffic
    
    d = d.drop_duplicates()
    d = d[feature_cols + META_KEEP].copy()      # 52 features + metadata

    # files in DROP and files with too few flows are skipped
    if len(d) < MIN_FLOWS:
        print(f"Skipped {cat}/{sub}: only {len(d)} flows after cleaning")
        continue
    d["category"] = cat
    d["subtype"] = sub
    if len(d) > CAP:
        d = d.sample(n=CAP, random_state=42)
    v, t = train_test_split(d, test_size=0.5, random_state=42)
    val_parts.append(v)
    test_parts.append(t)

attack_val = pd.concat(val_parts, ignore_index=True)
attack_test = pd.concat(test_parts, ignore_index=True)

# shows how many flows each attack contributes to validation
print(attack_val.groupby(["category", "subtype"]).size())
```

```python
# Create labels for benign data
for d in (benign_train, benign_val, benign_test):
    d["category"] = "Benign"
    d["subtype"] = "Benign"

# Create dataset for each train, velidation, and test
train_df = benign_train
val_df = pd.concat([benign_val, attack_val], ignore_index=True)
test_df = pd.concat([benign_test, attack_test], ignore_index=True)
```

```python
# Wrap preprocessing steps in a function
def clean(df):
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.dropna(subset=feature_cols).reset_index(drop=True)   # check features
    df["y"] = (df["category"] != "Benign").astype(int)           # 0 for Benign, 1 for Attack
    return df

train_df = clean(train_df)
val_df = clean(val_df)
test_df = clean(test_df)
```

```python
X_train = train_df[feature_cols]
X_val, y_val = val_df[feature_cols], val_df["y"].values
X_test, y_test = test_df[feature_cols], test_df["y"].values

print("Train:", X_train.shape, "| attacks:", train_df["y"].sum())     # 0
print("Val:  ", X_val.shape, "| benign/attack:", np.bincount(y_val))  # both are larger than 0
print("Test: ", X_test.shape, "| benign/attack:", np.bincount(y_test))
print(X_train.dtypes.value_counts())                                  # only numeric values
```

## Training an Isolation Model

```python
from sklearn.ensemble import IsolationForest
```

```python
model = IsolationForest(
    n_estimators=200, # num of trees. more stable with more trees (usually between 100-300)
    random_state = 42,
    n_jobs=-1, # Use all CPUs
)

model.fit(X_train)
```

### Return figures representing the model's quality
- ROC-AUC
- PR-AUC
- Detection rate of attacks @ 1% of false alarm rate
- detection rate by attacks

```python
from sklearn.metrics import roc_auc_score, average_precision_score
```

```python
val_scores = -model.score_samples(X_val) ## Make values large when it's suspecious

log = []

''' Display measures to see the model's quality. '''
def evaluate(name, scores, y, df, fpr=0.01):
    """ scores: Larger = more suspecious """
    th = np.percentile(scores[y == 0], 100 * (1 - fpr))
    alert = scores > th
    result = {
        "Name": name,
        "ROC-AUC": round(roc_auc_score(y, scores), 3),
        "PR-AUC" : round(average_precision_score(y, scores), 3),
        f"Detection Rate @ FPR {fpr:.0%}": round(alert[y == 1].mean(), 3) 
    }

    # Add result to log
    log.append(result)

    per = (df.assign(alert=alert)[y == 1].groupby("subtype")["alert"].mean().sort_values(ascending=False).round(3))
    print(result); print(per)
    return pd.DataFrame(log)

evaluate("Baseline", val_scores, y_val, val_df)
```

### Convert the result into log

```python
def signed_log(X):
    return np.sign(X) * np.log1p(np.abs(X))   # adapt to negative values

model_log = IsolationForest(n_estimators=200, random_state=42, n_jobs=-1)
model_log.fit(signed_log(X_train))
val_scores_log = -model_log.score_samples(signed_log(X_val))

evaluate("Baseline (log)", val_scores_log, y_val, val_df)
```
