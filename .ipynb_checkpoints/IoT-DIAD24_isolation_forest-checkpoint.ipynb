- Identify redundant columns and remove them

```python
sns.heatmap(benign24_b.corr())
```

```python
# Get negative-high correlation too
corr = benign24_b.corr().abs()

# Create a mask to keep an upper triangle
## np.ones - creates a matrix with ones in the same size as corr
## np.triu() - turn everything into 0 except for the upper triangle. k=1 allows diagonal to be 0
## .astype(bool) - Turn values into True/False
mask = np.triu(np.ones(corr.shape), k=1).astype(bool)

# Make False empty
upper_t = corr.where(mask)

# Turns its format into "pair: corr"
corr_pairs = upper_t.stack()

# Keep correlations when it's over 0.95
corr_pairs = corr_pairs[corr_pairs >= 0.95]

# Sort
corr_pairs = corr_pairs.sort_values(ascending=False)

corr_pairs
```

```python
# Define column lists to drop
## correlation=1.0 (both columns have the almost same info)
corr_1 = ['Fwd Segment Size Avg', 'Bwd Segment Size Avg', 'Average Packet Size']

## time + rate
time_rate = ['Fwd IAT Total', 'Fwd IAT Max', 'Idle Max', 'Idle Mean', 'Idle Min', 'Bwd IAT Min', 'Fwd Packets/s']

## the number of packets
n_packets = ['Fwd Header Length', 'Bwd Header Length', 'Bwd Packet/Bulk Avg', 'Bwd Bytes/Bulk Avg', 'Subflow Bwd Packets']

### NOTE
# The correlation between Total Bwd packets and Total Fwd packets is 0.99 as we receive a response to each request in normal traffic.
# However, it'll be broken when it is an attack such as SYN flood or port scan


h_corr_2drop = corr_1 + time_rate + n_packets
```

```python
benign24_b.drop(h_corr_2drop, axis = 1, inplace = True)
```

## Training model

### Train data

```python
from sklearn.model_selection import train_test_split
```

```python
benign_temp, benign_test = train_test_split(
    benign24_b,
    test_size=0.15,
    random_state = 42
)

display(benign_temp.shape, benign_test.shape)
```

```python
benign_train, benign_val = train_test_split(
    benign_temp,
    test_size=0.15,
    random_state=42
)
display(benign_temp.shape, benign_val.shape, benign_test.shape)
```

### Val, Test data

```python
attacks = {
    ("BruteForce", "DictionaryBruteForce"): bruteforce,
    ("DDoS", "ACK_Fragmentation"):  ddos_ack_frag,
    ("DDoS", "HTTP_Flood"):         ddos_http_flood,
    ("DDoS", "ICMP_Flood"):         ddos_icmp_flood,
    ("DDoS", "ICMP_Fragmentation"): ddos_icmp_frag,
    ("DoS", "HTTP_Flood"):          dos_http_flood,
    ("DoS", "SYN_Flood"):           dos_syn_flood,
    # ("DoS", "TCP_Flood"):         dos_tcp_flood,  # 除外：全5ファイルで1フローずつしかなく、攻撃の通信ではない
    ("DoS", "UDP_Flood"):           dos_udp_flood,
    ("Mirai", "Greeth_Flood"):      mirai_flood,
    ("Recon", "VulnerabilityScan"): recon,
    ("Spoofing", "DNS_Spoofing"):   spoofing_dns,
    ("Spoofing", "MITM_ArpSpoofing"): spoofing_mitm,
    ("Web-Based", "SqlInjection"):  web_sql_inj,
    ("Web-Based", "Uploading_Attack"): web_upload_attack,
    ("Web-Based", "XSS"):           web_xss,
}

for (cat, sub), d in attacks.items():
    print(f"{cat:10s} {sub:22s} {len(d):>9,} rows  columns={d.shape[1]}")
```

```python
# set the max number of rows for each attack file
CAP = 10000
META_KEEP = ["Src IP", "Dst IP", "Timestamp"]   # metadata to keep for analysis (not included in themodel)
val_parts, test_parts = [], []

for (cat, sub), d in attacks.items():
    d = d.drop_duplicates()                       # delete perfect duplicates before removing identifiers
    d = d[feature_cols + META_KEEP].copy()        # 54 features + metadata
    d["category"] = cat
    d["subtype"] = sub
    if len(d) > CAP:
        d = d.sample(n=CAP, random_state=42)
    v, t = train_test_split(d, test_size=0.5, random_state=42)
    val_parts.append(v)
    test_parts.append(t)

attack_val = pd.concat(val_parts, ignore_index=True)
attack_test = pd.concat(test_parts, ignore_index=True)
```

```python
# Create labels for benign data
for d in (benign_train, benign_val, benign_test):
    d["category"] = "Benign"
    d["subtype"] = "Benign"

# Create dataset for each train, velidation, and test
train_df = benign_train
val_df = pd.concat([benign_val, attack_val], ignore_index=True)
test_df = pd.concat([benign_test, attack_test], ignore_index=True)
```

```python
# Wrap preprocessing steps in a function
def clean(df):
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.dropna(subset=feature_cols).reset_index(drop=True)   # check features
    df["y"] = (df["category"] != "Benign").astype(int)           # 0 for Benign, 1 for Attack
    return df

train_df = clean(train_df)
val_df = clean(val_df)
test_df = clean(test_df)
```

```python
X_train = train_df[feature_cols]
X_val, y_val = val_df[feature_cols], val_df["y"].values
X_test, y_test = test_df[feature_cols], test_df["y"].values

print("Train:", X_train.shape, "| attacks:", train_df["y"].sum())     # 0
print("Val:  ", X_val.shape, "| benign/attack:", np.bincount(y_val))  # both are larger than 0
print("Test: ", X_test.shape, "| benign/attack:", np.bincount(y_test))
print(X_train.dtypes.value_counts())                                  # only numeric values
```

## Training an Isolation Model

```python
from sklearn.ensemble import IsolationForest
```

```python
model = IsolationForest(
    n_estimators=200, # num of trees. more stable with more trees (usually between 100-30)
    random_state = 42,
    n_jobs=-1, # Use all CPUs
)

model.fit(X_train)
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

val_scores = -model.score_samples(X_val) # .score_samples gives larger values if it's normal, so we flip it so that it means large if it's anomaly

print("ROC-AUC:", round(roc_auc_score(y_val, val_scores), 4))
print("PR-AUC: ", round(average_precision_score(y_val, val_scores), 4))
print("Pretty low score... T-T")
```

### Look at result by attacks

```python
val_df["score"] = val_scores
benign_s = val_df.loc[val_df["y"] == 0, "score"].values

rows = []
for (cat, sub), g in val_df[val_df["y"] == 1].groupby(["category", "subtype"]):
    y = np.r_[np.zeros(len(benign_s)), np.ones(len(g))]
    s = np.r_[benign_s, g["score"].values]
    rows.append((cat, sub, len(g), round(roc_auc_score(y, s), 3)))

per_attack = pd.DataFrame(rows, columns=["category", "subtype", "n", "ROC-AUC"])
print(per_attack.sort_values("ROC-AUC"))
```

```python
def signed_log(X):
    return np.sign(X) * np.log1p(np.abs(X))   # adapt to negative values

model_log = IsolationForest(n_estimators=200, random_state=42, n_jobs=-1)
model_log.fit(signed_log(X_train))
val_scores_log = -model_log.score_samples(signed_log(X_val))

print("ROC-AUC (log):", round(roc_auc_score(y_val, val_scores_log), 4))
```

```python
cols_no_port = [c for c in feature_cols if c != "Dst Port"]

model_np = IsolationForest(n_estimators=200, random_state=42, n_jobs=-1)
model_np.fit(X_train[cols_no_port])
s = -model_np.score_samples(X_val[cols_no_port])
print("ROC-AUC (no port):", round(roc_auc_score(y_val, s), 4))
```

```python
# Thresholds from benign validation scores (higher score = more anomalous)
benign_val_scores = val_scores_log[y_val == 0]
warn_threshold = np.percentile(benign_val_scores, 99)    # ~1% of normal flows get WARNING or above
flag_threshold = np.percentile(benign_val_scores, 99.9)  # ~0.1% of normal flows get FLAGGED

alerted = val_scores_log > warn_threshold
print(f"False alarm rate (benign): {alerted[y_val == 0].mean():.2%}")
print(f"Detection rate (attacks):  {alerted[y_val == 1].mean():.2%}")

# Detection rate per attack type
val_df["alerted"] = alerted
print(val_df[val_df["y"] == 1]
      .groupby(["category", "subtype"])["alerted"].mean()
      .sort_values(ascending=False).round(3))
```
